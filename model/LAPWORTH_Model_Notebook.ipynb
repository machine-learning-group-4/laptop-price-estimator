```python
#Importing sklearn libraries
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
```

```python
#Identifying numerical and categorical columns
numerical_columns = ["Inches", "Ram", "Weight"]

categorical_columns = [
    "Company",
    "TypeName",
    "ScreenResolution",
    "Cpu",
    "Memory",
    "Gpu",
    "OpSys"
]
```

```python
#Encoding categorical columns using onehotencoder
preprocessor = ColumnTransformer(
    transformers=[
        ("num", SimpleImputer(strategy="median"), numerical_columns),
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_columns)
    ]
)
```

```python
#Creaing the model using Random Forest mode
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        random_state=42
    ))
])
```

```python
#Model training
model.fit(X_train, y_train)
```

```python
print("Model training completed successfully!")
```

```python
#Saving trained model
import joblib

joblib.dump(model, "laptop_price_model.pkl")
```
